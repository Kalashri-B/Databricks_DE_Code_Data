### Display analysis of current and historical data

```python
%run "../00_common/01_environment_config"
```

```python
from pyspark.sql import functions as F
```

```python
# Read our reporting dataset

driver_path = f"{s3_root_path}/Gold/Kalashri/driver_performance"
constructor_path = f"{s3_root_path}/Gold/Kalashri/constructor_performance"

driver_df = spark.read.format("delta").load(driver_path)

constructor_df = spark.read.format("delta").load(constructor_path)
```

```python
# Find latest season in our data

latest_season = driver_df.agg(
    F.max("season")
).first()[0]

print("Latest Season:", latest_season)
```

```python
# Current season driver performance

current_driver_df = (
    driver_df
    .filter(F.col("season") == latest_season)
    .orderBy(F.desc("total_points"))
)

display(current_driver_df)
```

```python
# Historical driver performace

historical_driver_df = (
    driver_df
    .orderBy(
        "driver_name",
        "season"
    )
)

display(historical_driver_df)
```

```python
# Current season constructor performance

current_constructor_df = (
    constructor_df
    .filter(F.col("season") == latest_season)
    .orderBy(F.desc("total_points"))
)

display(current_constructor_df)
```

```python
# Historical constructor performance

historical_constructor_df = (
    constructor_df
    .orderBy(
        "constructor_name",
        "season"
    )
)

display(historical_constructor_df)
```
