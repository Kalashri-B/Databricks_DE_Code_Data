## dbutils


A collection of built-in Databricks utilities used inside notebooks to work with files, parameters, secrets, notebooks, and jobs.

### 1. dbutils.fs — File System Utilities

dbutils.fs is used to work with files and folders.

```python
# dbutils.fs.ls()	       → dbutils.fs.ls("path")	                     → Lists files and folders available at a given path.
# dbutils.fs.put()	       → dbutils.fs.put("path", "content", True)	 → Creates a file and writes text into it. True overwrites the existing file.
# dbutils.fs.cp()	       → dbutils.fs.cp("source", "destination")      → Copies a file from one location to another.
# dbutils.fs.mv()	       → dbutils.fs.mv("source", "destination")	     → Moves a file from one location to another.
# dbutils.fs.rm()	       → dbutils.fs.rm("path")	                     → Deletes a file.
# dbutils.fs.mkdirs()      → dbutils.fs.mkdirs("path")	                 → Creates a new directory/folder.
# dbutils.fs.head()	       → dbutils.fs.head("path", maxBytes)           → Reads the beginning portion of a file.
```

### 2. dbutils.widgets — Notebook Parameters

It makes notebooks dynamic instead of hardcoding values.

```python
# dbutils.widgets.text()	       → dbutils.widgets.text("name", "default_value")	            → Creates a text parameter in a notebook.
# dbutils.widgets.get()	           → dbutils.widgets.get("name")	                            → Gets the value of a parameter so we can use it in our code.
# dbutils.widgets.dropdown()	   → dbutils.widgets.dropdown("name", "default", ["A", "B"])	→ Creates a dropdown parameter when values should be selected from predefined options.
# dbutils.widgets.removeAll()	   → dbutils.widgets.removeAll()	                            → Removes all widgets/parameters from the notebook.
```
