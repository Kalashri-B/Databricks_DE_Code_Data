```python

# MAGIC %run ../00_common/01_environment_config

dbutils.widgets.text("p_batch_id", "")
v_batch_id = dbutils.widgets.get("p_batch_id").strip()

control_table_path = (
    f"{s3_root_path}/Control/Kalashri_increment/batch_control"
)

from pyspark.sql import Row
from pyspark.sql import functions as F

if not v_batch_id:
    raise Exception("batch_id is missing")

in_progress_df = (
    spark.createDataFrame([
        Row(batch_id=v_batch_id, status="in_progress")
    ])
    .withColumn("created_timestamp", F.current_timestamp())
    .withColumn("updated_timestamp", F.current_timestamp())
)

(
    in_progress_df.write
    .format("delta")
    .mode("append")
    .save(control_table_path)
)

print(f"Marked batch {v_batch_id} as in_progress")
```
