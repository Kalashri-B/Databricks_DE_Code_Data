```python

# Databricks notebook source

# ============================================================
# F1 PROJECT - CREATE CONTROL TABLES
# ============================================================
# Purpose:
# Create the batch_control Delta table in S3
# for incremental-load orchestration.
# ============================================================

# 1. Load existing environment configuration
# MAGIC %run ../00-common/01.environment-config

from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    TimestampType
)

# 2. Define the control table S3 path
control_table_path = (
    f"{s3_root_path}/Control/Kalashri_increment/batch_control"
)

# 3. Define the batch_control schema
control_table_schema = StructType([
    StructField("batch_id", StringType(), True),
    StructField("status", StringType(), True),
    StructField("created_timestamp", TimestampType(), True),
    StructField("updated_timestamp", TimestampType(), True)
])

# 4. Create an empty Delta table if it does not exist
empty_df = spark.createDataFrame([], control_table_schema)

(
    empty_df.write
    .format("delta")
    .mode("ignore")
    .save(control_table_path)
)

print(f"Control table ready at: {control_table_path}")
```

```python
display(
    spark.read
    .format("delta")
    .load(control_table_path)
)
```
