```python
# ============================================================
# Master Silver Notebook - F1 Incremental Load
# Purpose: Run all Silver transformation notebooks for a batch
# ============================================================

# 1. Get batch ID
dbutils.widgets.text("p_batch_id", "")

v_batch_id = dbutils.widgets.get("p_batch_id").strip()

if not v_batch_id:
    raise Exception("p_batch_id is missing. Please provide a valid batch ID.")

print(f"Starting Silver transformations for batch: {v_batch_id}")


# 2. List of Silver transformation notebooks
silver_notebooks = [
    "01_transform_circuits_data",
    "02_transform_races_data",
    "03_transform_constructors_data",
    "04_transform_drivers_data",
    "05_transform_results_data",
    "06_transform_sprints_data"
]


# 3. Execute notebooks sequentially
for notebook in silver_notebooks:
    print(f"Starting Silver transformation: {notebook}")

    try:
        result = dbutils.notebook.run(
            f"./{notebook}",
            timeout_seconds=0,
            arguments={"p_batch_id": v_batch_id}
        )

        print(f"Completed Silver transformation: {notebook}")

    except Exception as e:
        print(f"FAILED Silver transformation: {notebook}")
        print(f"Batch ID: {v_batch_id}")
        raise


# 4. Completion message
print(f"All Silver transformations completed successfully for batch: {v_batch_id}")
```
