```python

# ============================================================
# Master Gold Notebook - F1 Incremental Load
# Purpose: Run all Gold transformation notebooks for a batch
# ============================================================

# 1. Get batch ID
dbutils.widgets.text("p_batch_id", "")
v_batch_id = dbutils.widgets.get("p_batch_id").strip()

if not v_batch_id:
    raise Exception("p_batch_id is missing. Please provide a valid batch ID.")

print(f"Starting Gold transformations for batch: {v_batch_id}")


# 2. Base path for Gold notebooks
gold_path = (
    "/Workspace/Users/kalashri.borhade@bizmetric.com/"
    "Databricks_DE_Code_Data/F1_incremental_load/04_gold"
)

# 3. Gold transformation notebook paths
gold_notebooks = [
    "01_buils_races_dimension",
    "02_build_constructors_dimension",
    "03_build_drivers_dimension",
    "04_build_results_fact"
]


# 4. Execute notebooks sequentially
for notebook in gold_notebooks:
    notebook_path = f"{gold_path}/{notebook}"

    print(f"Starting Gold transformation: {notebook}")
    print(f"Notebook path: {notebook_path}")

    try:
        result = dbutils.notebook.run(
            notebook_path,
            timeout_seconds=0,
            arguments={"p_batch_id": v_batch_id}
        )

        print(f"Completed Gold transformation: {notebook}")
        print(f"Result: {result}")

    except Exception as e:
        print(f"FAILED Gold transformation: {notebook}")
        print(f"Batch ID: {v_batch_id}")
        print(f"Error: {str(e)}")
        raise


# 5. Completion message
print(
    f"All Gold transformations completed successfully "
    f"for batch: {v_batch_id}"
)
```
