```python
# Databricks notebook source

# ============================================================
# F1 PROJECT - CREATE SCHEMAS
# ============================================================
# Purpose:
# Creates the Bronze, Silver, and Gold schemas required for
# the F1 Data Engineering project.
# ============================================================

# COMMAND ----------

catalog_name = "bizmetric_databricks_learning"

# COMMAND ----------

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {catalog_name}.bronze
""")

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {catalog_name}.silver
""")

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {catalog_name}.gold
""")

# COMMAND ----------

spark.sql(f"SHOW SCHEMAS IN {catalog_name}").show()
```
