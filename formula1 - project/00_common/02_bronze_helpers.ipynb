```python
# Databricks notebook source

# ============================================================
# F1 PROJECT - BRONZE HELPER FUNCTIONS
# ============================================================
# Purpose:
# Contains reusable functions used during Bronze ingestion.
# Adds ingestion metadata to identify when and from which
# source file the data was loaded.
# ============================================================

# COMMAND ----------

from pyspark.sql import functions as F


# COMMAND ----------

def add_ingestion_metadata(df):

    return (
        df.withColumn("ingestion_timestamp", F.current_timestamp())
          .withColumn("source_file",F.col("_metadata.file_path")))
```
